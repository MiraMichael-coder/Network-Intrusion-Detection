```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

from sklearn.dummy import DummyClassifier
from sklearn.pipeline import make_pipeline
from src.config import CAT_COLS
from src.data import load_train
from src.evaluate import cv_evaluate
from src.preprocessing import create_preprocessor

train = load_train()
X = train.drop(columns=["id", "attack_cat", "label"])
y = train["label"]
```

```python
num_cols = [c for c in X.columns if c not in CAT_COLS]
pipe = make_pipeline(create_preprocessor(num_cols), DummyClassifier(strategy="most_frequent"))
cv_evaluate("dummy_most_frequent", pipe, X, y)
```
